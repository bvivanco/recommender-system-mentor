### Transformar skills en una Matriz

```python
df_globers.columns
```

```python
def skills_to_vector(row):
    hard_skills = eval(row['skills'])['hard']  # Usar eval para convertir la cadena a diccionario
    soft_skills = eval(row['skills'])['soft']
    
    # Crear un diccionario de habilidades y puntuaciones
    skill_dict = {}
    
    # Procesar habilidades duras
    for skill in hard_skills:
        skill_name, score = skill.split(': ')
        skill_dict[skill_name.strip()] = int(score.strip())
    
    # Procesar habilidades blandas
    for skill in soft_skills:
        skill_name, score = skill.split(': ')
        skill_dict[skill_name.strip()] = int(score.strip())
    
    return pd.Series(skill_dict)

# Aplicar la función al DataFrame
skills_matrix = df_globers.apply(skills_to_vector, axis=1).fillna(0)

# Mostrar la matriz de habilidades
print(skills_matrix.head())
```

```python
from sklearn.model_selection import train_test_split
import numpy as np

# Crear una matriz de interacciones (simulación de puntuaciones)
interactions = np.random.randint(1, 6, size=skills_matrix.shape)  # Simulando puntuaciones entre 1 y 5

# Dividir en conjunto de entrenamiento y prueba
train_data, test_data = train_test_split(interactions, test_size=0.2, random_state=42)
```

### Implementar y entrenar Funk-SVD 

```python
class FunkSVD:
    def __init__(self, n_factors=10, n_epochs=1000, lr=0.01, reg=0.02):
        self.n_factors = n_factors
        self.n_epochs = n_epochs
        self.lr = lr
        self.reg = reg

    def fit(self, train_data):
        n_users, n_items = train_data.shape
        self.user_factors = np.random.normal(scale=1./self.n_factors, size=(n_users, self.n_factors))
        self.item_factors = np.random.normal(scale=1./self.n_factors, size=(n_items, self.n_factors))

        for epoch in range(self.n_epochs):
            for i in range(n_users):
                for j in range(n_items):
                    if train_data[i][j] > 0:
                        error = train_data[i][j] - self.predict(i, j)
                        self.user_factors[i] += self.lr * (error * self.item_factors[j] - self.reg * self.user_factors[i])
                        self.item_factors[j] += self.lr * (error * self.user_factors[i] - self.reg * self.item_factors[j])

    def predict(self, user_id, item_id):
        return np.dot(self.user_factors[user_id], self.item_factors[item_id])

# Entrenamiento del modelo
model = FunkSVD(n_factors=10)
model.fit(train_data)
```

### Evaluar el modelo 🚀

```python
from sklearn.metrics import mean_squared_error

def evaluate_model(model, test_data):
    predictions = []
    actuals = []
    
    for i in range(test_data.shape[0]):
        for j in range(test_data.shape[1]):
            if test_data[i][j] > 0: 
                predictions.append(model.predict(i, j))
                actuals.append(test_data[i][j])
    
    rmse = np.sqrt(mean_squared_error(actuals, predictions))
    mae = np.mean(np.abs(np.array(predictions) - np.array(actuals)))
    
    return rmse, mae

rmse, mae = evaluate_model(model, test_data)
print(f"RMSE: {rmse}, MAE: {mae}")
```
